```python
data.isnull().sum()
```

```python
print(data['Age'].mean())


data['Age']=data['Age'].fillna(data['Age'].mean(),inplace=True)

data.isnull().sum()
```

```python
data['Embarked'] = data['Embarked'].fillna('S')

data.isnull().sum()
```

```python
data.drop(columns=['Cabin'], inplace =True)

data.isnull().sum()
```

```python
#feature engineering is a process of creating , transforming or extracting info from the features before making ml model ready 
data.head(10)
```

```python
data.columns
```

```python
#get_dummies
pclass = pd.get_dummies(data['Pclass'],drop_first=True,dtype=int)
pclass
```

```python
#get_dummies
sex = pd.get_dummies(data['Sex'],drop_first=True,dtype=int)
sex
```

```python
#get_dummies
embarked = pd.get_dummies(data['Embarked'],drop_first=True,dtype=int)
embarked
```

```python
#Concatenation of all tables
final_data = pd.concat([data,pclass,sex,embarked], 
                       axis = 1)
final_data
```

```python
final_data.drop(columns=['PassengerId','Name','Pclass','Sex','Ticket','Embarked'],inplace=True)
final_data.head()
```
